# Class projects: candidate papers for omics replication

Prepared by Christoforos Nikolaou for the Introduction to Biological Data Analysis class (groups of 4-5 students, analysis in R)

## How to use this list

Each entry gives the paper link, the data link, a short summary (main question and findings) and introductory questions that help a group get into the topic. Notes on data formats are as supplied by the instructor; anything marked 'to check' has not been verified. Papers 6, 8, 9 and 10 are the most demanding.

## 1. Macrophage enhancer landscape in innate immune activation and tolerance

Genome Biology, 2022

- **Paper:** <https://link.springer.com/article/10.1186/s13059-022-02702-1>
- **Data:** GEO GSE172116, <https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE172116>. Accessible data in bigWig, bedGraph, tab-delimited files and similar.

**Summary**

- Question: how does the macrophage enhancer repertoire differ between the acute endotoxin response and endotoxin tolerance?
- Design: ATAC-seq, H3K27ac and H3K4me3 ChIP-seq and total RNA-seq in human macrophages, linked to splicing, TF binding and enhancer RNA signatures.
- Findings: tolerance-associated enhancers are enriched for context-specific eQTLs and for GWAS variants in inflammatory bowel disease and multiple sclerosis; stem cell-derived macrophages are compared with primary cells, with enhancer activity tested at KLF4, SLAMF1 and IL2RA.

**Introductory questions**

- What is the difference between a naive, an acutely stimulated and a tolerant macrophage, and how would each show up in RNA-seq?
- Which histone marks separate promoters from active enhancers, and how do you decide that an ATAC peak is an enhancer?
- Can you reproduce a PCA of the samples and the clustering of the differential regions?

## 2. The epigenetic state of IL-4-polarized macrophages enables inflammatory cistromic expansion and extended synergistic response to TLR ligands

Czimmerer et al., Immunity, 2022

- **Paper:** <https://www.cell.com/immunity/fulltext/S1074-7613(22)00545-3>
- **Data:** well documented supplementary files inside the paper itself.

**Summary**

- Question: IL-4 and TLR programs are thought to antagonize each other, so what happens when IL-4-polarized macrophages meet LPS?
- Finding: IL-4-polarized macrophages mount a hyperinflammatory program after LPS exposure (extended synergy), supported by epigenomic remodeling, expansion of the NF-kB p65 cistrome and increased enhancer activity.
- Finding: STAT6- and EGR2-deficient macrophages lose much of the effect, and BRD4 inhibition reduces it.

**Introductory questions**

- Why would a classical and an alternative activation program be expected to oppose each other?
- Which genes show super-induction after priming, and can you define synergy quantitatively (observed versus additive expectation)?
- Do the newly gained p65 peaks sit in pre-accessible regions or in newly opened ones?

## 3. Subtype-specific 3D genome alteration in acute myeloid leukaemia

Nature, 2022

- **Paper:** <https://doi.org/10.1038/s41586-022-05365-x>
- **Data:** GEO GSE152136, <https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE152136>. See Custom Downloads in the TAR ball.

**Summary**

- Question: how do A/B compartments, TADs and loops change across AML subtypes, and can hypomethylating agents restore normal organization?
- Finding: genes in regions switching from A to B or from B to A show decreased or increased expression, respectively.
- Finding: most TAD alterations do not change gene expression; the study also relates DNA methylation and CTCF binding to 3D structure.

**Introductory questions**

- What are the A and B compartments, and what does a PC1 value mean?
- Do genes in switching regions change expression in the expected direction? Reproduce the boxplot.
- Why might most TAD changes have no expression effect, and what does that imply for the idea that structure determines function?

## 4. Gene architecture directs splicing outcome in separate nuclear spatial regions

Tammer et al., Molecular Cell, 2022

- **Paper:** <https://cell.com/molecular-cell/retrieve/pii/S109727652200106X>
- **Data:** supplementary data available with the paper.

**Summary**

- Question: how does the splicing machinery decide whether the exon or the intron is the unit to define?
- Finding: peripheral genes have low-GC exons flanked by long, lower-GC introns and are spliced by exon definition (exon skipping when splicing goes wrong); central genes have GC-rich exons and short flanking introns and are spliced mainly by intron definition (intron retention when splicing goes wrong).
- Finding: two sets of splicing factors form separate regulatory subnetworks for the two gene architectures.

**Introductory questions**

- What are exon definition and intron definition, and why would intron length matter?
- Classify genes by exon and intron GC content. Do the classes differ in splicing outcome and in nuclear position (lamina, compartment)?
- How do isochores relate to this gene architecture?

## 5. Engineered MED12 mutations drive leiomyoma-like transcriptional and metabolic programs by altering the 3D genome compartmentalization

Nature Communications, 2023

- **Paper:** <https://www.nature.com/articles/s41467-023-39684-y>
- **Data:** GEO GSE226017, <https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE226017>. See Custom Downloads in the TAR ball.

**Summary**

- Question: do MED12 hotspot mutations found in leiomyoma reprogram transcription through changes in compartments?
- Finding: the mutation causes genome-wide A/B reorganization, with roughly 194 Mb switching from B to A and 2.6 Gb from A to B.
- Finding: expression changes are compared for genes in B-to-A, stable and A-to-B regions.

**Introductory questions**

- Why use an engineered isogenic mutant rather than patient tissue?
- What fraction of differentially expressed genes lie in switching regions, and is that more than chance (permutation test)?
- Do switching regions differ in GC content or gene density from stable ones?

## 6. Nuclear speckles and the way GC%-rich mRNA is processed (Aktas lab)

Malszycki et al., Cell, 2026: Nuclear speckles enable processing of RNA from GC-rich isochores

- **Paper:** <https://www.cell.com/cell/fulltext/S0092-8674(26)00098-X>
- **Data:** to check.
- **Concepts:** GC%, isochores, splicing, evolution of gene structure.

**Summary**

- Question: what is the function of nuclear speckles?
- Finding: acute removal of the speckle core proteins SON and SRRM2 downregulates GC-rich genes with short introns that cluster in GC-rich isochores, through inefficient and chaotic splicing; genes outside these isochores are unaffected.
- Finding: GC-rich isochores and speckles are found only in amniotes, and the disordered regions of SON expanded in that lineage, suggesting co-evolution.

**Introductory questions**

- What is an isochore, and why might GC-rich, gene-dense regions be hard to splice?
- Which genes go down after SON/SRRM2 depletion, and what do they share (GC3, intron length, isochore family)?
- Why would the co-occurrence of speckles and isochores in amniotes point to co-evolution rather than coincidence?

## 7. Increased enhancer-promoter interactions during developmental enhancer activation in mammals

Nature Genetics, 2024

- **Paper:** <https://www.nature.com/articles/s41588-024-01681-2>
- **Data:** GEO GSE217078, <https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE217078>. CTCF data (bigWig and peaks) in GSM5501396, GSM5501397 and GSM5501398.

**Summary**

- Question: does enhancer activation involve increased physical proximity to the target promoter?
- Design: enhancer-capture Hi-C for 935 developmental enhancers across 10 embryonic mouse tissues; interaction calls are given in a supplementary table.
- Finding: most enhancers have tissue-specific 3D conformations, and enhancer-promoter and enhancer-enhancer interactions increase moderately but consistently upon activation.

**Introductory questions**

- How do you define an active versus a poised enhancer from H3K27ac and H3K4me1?
- From the interaction tables, are enhancer-promoter contacts stronger in the tissue where the enhancer is active, and how big is the effect?
- Is moderately but consistently higher enough to claim causality?

## 8. A single-cell massively parallel reporter assay detects cell-type-specific gene regulation

Zhao et al., Nature Genetics, 2023 (harder)

- **Paper:** <https://www.nature.com/articles/s41588-022-01278-7>
- **Data:** GEO GSE188639, <https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE188639>. Some data here; not checked entirely.

**Summary**

- Question: can reporter assays be made cell-type-specific without running them serially in each cell type?
- Finding: a library of core promoters assayed in a mix of HEK293 and K562 cells detects cell-type-specific activity, with reporter copy number normalized per cell.
- Finding: promoter variants measured in live mouse retina show cell-type-specific effects of subtle variants.

**Introductory questions**

- What is an MPRA, and why is barcode counting the key readout?
- Which core promoters differ between K562 and HEK293, and do they differ in GC content or motif content?
- Why is normalizing for per-cell plasmid copy number necessary?

## 9. Senescent cells cluster CTCF on nuclear speckles to instruct an alternative splicing program

Palikyras et al., Nature Aging, 2026 (harder)

- **Paper:** <https://www.nature.com/articles/s43587-026-01171-6>
- **Data:** various GEO accession numbers. One found so far: GSE269857, an eCLIP dataset listed for the preprint version (<https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE269857>); the others are to be listed.

**Summary**

- Question: why does CTCF form large clusters (SICCs) in senescent cells, and do they have a function?
- Finding: cells entering senescence repurpose SRRM2 and BANF1 to cluster CTCF on the surface of speckles while rewiring higher-order genome architecture.
- Finding: disrupting SICCs by SRRM2 or BANF1 depletion reverts alternative splicing choices.

**Introductory questions**

- Which events mark senescence, and how does a senescent transcriptome differ from a quiescent one?
- From the splicing tables, which alternative splicing event types change most, and do they follow the flat-exon pattern?
- Do the changed events lie near speckle-associated regions? This is a natural bridge to papers 4 and 6.

## 10. Position-dependent function of human sequence-specific transcription factors

Duttke et al., Nature, 2024 (harder)

- **Paper:** <https://www.nature.com/articles/s41586-024-07662-z>
- **Data:** no data link supplied yet; to check.

**Summary**

- Question: does where a TF binds relative to the transcription start site change what it does?
- Finding: many TFs, including canonical activators such as NRF1, NFY and Sp1, activate or repress initiation depending on their position relative to the TSS.
- Finding: motif enrichment near TSSs shows about 10 bp periodicity, suggesting rotational positioning matters.

**Introductory questions**

- What are the textbook definitions of activator and repressor, and which results here challenge them?
- Plot motif density around the TSS for a few TFs. Where is the preferred position, and does GC content bias confound it?
- What would a regression of expression change on binding position upstream versus downstream predict?

## 11. Senescence-related inflammation driven by transcription of repetitive elements

Neretti and Sedivy labs, Nature (published 2019 as: L1 drives IFN in senescent cells and promotes age-associated inflammation)

- **Paper:** <https://www.nature.com/articles/s41586-018-0784-9>
- **Companion paper:** Regulation of human interferon signaling by transposon exonization, Cell 2024, <https://doi.org/10.1016/j.cell.2024.11.016>. Supplementary Table 1 may be ok.
- **Data:** to check.
- **Concepts:** senescence, inflammation, mobile elements, L1 transcription, (de)-heterochromatinization.

**Summary**

- Question: what links cellular senescence to sterile, age-associated inflammation?
- Finding: during senescence L1 retrotransposons become transcriptionally derepressed and activate a type I interferon response, triggered by cytoplasmic L1 cDNA and antagonized by L1 reverse transcriptase inhibitors.
- Companion: transposon exonization generates alternative isoforms of immune genes, including a short IFNAR2 isoform that acts as a decoy receptor and dampens interferon signaling.

**Suggested questions**

- How did the authors identify the candidate factors affecting L1 transcription? Can you try to do this yourselves (ENCODE, bedtools and so on)?
- Link this with the Cell 2024 paper: how many immune genes carry transposon-derived exons, and are they enriched for particular TE families?
- How do you quantify repeat expression when reads map to many near-identical copies?

## Notes and open points

- Supplement file formats (csv, tsv, txt versus xlsx) have not yet been verified for any of the papers.
- Paper 11 was published in 2019, outside the 5-year window, and its data availability should be checked; the title in the listing differs from the published title.
- Papers 4, 6 and 9 form a natural themed cluster (GC architecture, speckles, splicing), which allows comparison between groups.
- For papers 8 and 10, data links and processed tables are the main open question.
